```python
#Example
def evenOdd(x:int):
    if(x%2==0):
        return "Even"
    else:
        return "Odd"

print(evenOdd(16))
print(evenOdd(7))  
```

```python
#Default Argument
def myFun(x, y=50):
    print("x: ",x)
    print("y: ",y)

myFun(10)

#here y is an argument(value) and x is a parameter(variable)
```

```python
#Default Argument
def myFun(country="Norway"):
    print("I am from "+country)

myFun("India")
myFun("Sweden")
myFun()
myFun("Brazil")

#here the default is Norway, but passing another argument overwrites the default 
```

```python
#Keyword Argument
def student(firstname, lastname):
    print(firstname, lastname)

student(firstname="Tony", lastname="Stark")
student(lastname="Stark", firstname="Tony")

#since we are assigning values to the keyword, interchanging the order doesn't change the order of the output as long as we have our desired input on the approriate keyword.
```

```python
#Position Arguments
def nameAge(name, age):
    print("Hi! I am ", name)
    print("My age is ", age)
print("Case1: ")
nameAge("Suraj", 27)
print("Case2: ")    
nameAge(27, "Suraj")
```
